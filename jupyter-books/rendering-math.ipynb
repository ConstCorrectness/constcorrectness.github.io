# 3D Projection Mathematics

To render a 3D object onto a 2D screen (like in our React Three Fiber viewer), we use projection matrices. Here is how the math works in Python.

```python
import numpy as np
import matplotlib.pyplot as plt

def perspective_projection(point_3d, d=1):
    x, y, z = point_3d
    return np.array([x / (z/d), y / (z/d)])

# Define a simple 3D cube vertex
point = np.array([10, 10, 20])
projected = perspective_projection(point)

print(f"3D Point: {point} -> 2D Screen: {projected}")
```
